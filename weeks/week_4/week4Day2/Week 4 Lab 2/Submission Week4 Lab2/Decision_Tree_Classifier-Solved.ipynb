#Decision Tree Classifier Lab


#Load essential libraries

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn import tree
from sklearn.model_selection import train_test_split
```

#Import the dataset

```python
df = pd.read_csv("titanic.csv")
df
```

##Drop meaningless information

```python
df = df.drop(columns = ['PassengerId','Name','SibSp', 'Parch', 'Ticket', 'Cabin', 'Embarked'], axis = 'columns')
df
```

##Data Preprocessing

```python
df.isna().sum()
```

```python
import math
mean_age = df.Age.mean()
mean_age = math.floor(mean_age)
print(mean_age)
```

```python
df['Age'] = df.Age.fillna(mean_age)
df
```

##Label Encode the non numeric features

```python
LE_sex = LabelEncoder()
df['Sex'] = LE_sex.fit_transform(df['Sex'])

df
```

## Define Feature Matrix (X) and Target Vector (y)


```python
y = df['Survived']
X = df.drop('Survived', axis = 'columns')
X
```

```python
df
```

## Train-Test Split and Model Initialization


```python
model = tree.DecisionTreeClassifier()

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.4)
```

##Train the model

```python
model.fit(X_train, y_train)

model.score(X_test, y_test)
#model.score returns the mean accuracy for test data and labels
```

```python
X_test
```

##Predict results

```python
model.predict([[1,0,19.0,91.0792]])
```
